## 3. Success metric

*One metric you can defend. What number means 'good'?*


Metric: Macro F1-score (with special attention to recall on the "down" class)

Accuracy is a poor choice here because the classes are imbalanced — from my Week 1 groupby, "down" has ~16,262 pages while "flat" has only ~1,152. A model could score high accuracy just by always predicting "down" and ignoring the smaller classes entirely, which would be useless in practice. Macro F1 treats each class equally regardless of size, so it forces the model to actually perform reasonably across all trend categories, not just the dominant one. I'll also track recall specifically on "down", since missing a genuinely declining page (a false negative) is the costliest mistake for the action this supports — a content team needs to actually catch declining pages, not just be broadly "accurate."
